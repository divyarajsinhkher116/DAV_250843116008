#Practical 2:-Perform data inspection, cleaning, handling missing values, duplicate removal, and data transformation using Pandas.

Step1: Import Pandas

```python
## Import pandas library
import pandas as pd
```

Step2: Create Dataframe

```python
# Create data dictionary
data ={
    "Id":[101,102,103,104,104,105,106,107],
    "Name":["Rahul","Priya","Amit",None,"Riya","Ajay","Naresh","Jainil"],
    "Age":[20,21,None,20,20,22,23,28],
    "City":["Surat","Navsari","Surat","Bharuch","Bharuch","Navsari","Surat","Ahemdabad"],
    "Marks":[85,None,78,88,88,81,90,93]
}
```

```python
# Convert to DataFrame
df = pd.DataFrame(data)
df
```

```python
# Display first rows
df.head()
```

```python
# Display last rows
df.tail()
```

```python
# Show DataFrame info
df.info()
```

```python
# Get statistical info
df.describe()
```

```python
# Show DataFrame shape
df.shape
```

```python
# Display column names
df.columns
```

```python
# Show column types
df.dtypes
```

```python
# Check missing values
df.isnull().sum()
```

```python
# Fill Marks NaNs
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df
```

```python
# Fill Age NaNs
df["Age"] = df["Age"].fillna(df["Age"].mean())
df
```

```python
# Fill Name NaNs
df["Name"] = df["Name"].fillna("Unknown")
df
```

```python
# Detect duplicates
df.duplicated()
```

```python
# Count duplicates
df.duplicated().sum()
```

```python
# Remove duplicates
df = df.drop_duplicates()
df
```

```python
# Rename 'Marks' column
df.rename(columns ={"Marks":"Score"},inplace=True)
df
```

```python
# Convert Age to int
df["Age"] = df["Age"].astype(int)
df.dtypes
```

```python
# Create Final Score
df["Final Score"] = df["Score"]+5
df
```

```python
# Uppercase Name column
df["Name"] = df["Name"].str.upper()
df
```

```python
# Lowercase City column
df["City"] = df["City"].str.lower()
df
```

```python
# Strip City whitespace
df["City"] = df["City"].str.strip()
df
```

```python
# Sort by Final Score
df_sorted = df.sort_values(by="Final Score",ascending=False)
df_sorted
```

```python
# Save to CSV
df.to_csv("Cleaned_Students.CSV",index=False)
print("Dataset created successfully")
```
