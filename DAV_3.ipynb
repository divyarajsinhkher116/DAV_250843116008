##Practical 3 :- Apply ƒltering, sorting, aggregation, and GroupBy operations on real-world datasets.

step 1 Create and Display DataFrame

```python
# DataFrame initialization.
import pandas as pd

data = {
    'ID': [100, 101, 102, 103, 104, 105, 106, 107,],
    'Name': ['Rahul', 'Priya', 'Amit', 'Karan', 'Neha', 'pooja', 'vivek', 'Riya'],
    'Dept': ['IT', 'HR', 'IT', 'Finance', 'HR', 'Finance', 'IT', 'HR'],
    'City': ['Surat', 'Navsari', 'Surat', 'Valsad', 'Surat', 'Navsari', 'Valsad', 'Surat'],
    'Salary': [45000, 38000, 52000, 48000, 40000, 51000, 60000, 42000],
    'Experience': [2, 1, 5, 4, 2, 6, 8, 3]
}

df = pd.DataFrame(data)
```

```python
# Display DataFrame.
df
```

step 2 Display IT Department Employees

```python
# Filter by IT department.
df[df["Dept"]=="IT"]
```

step 3 Display Employees with Salary Greater Than 50k

```python
# Filter by Salary > 50k.
df[df["Salary"]>50000]
```

step 4 Display Employees in Surat City

```python
# Filter by Surat city.
df[df["City"]=="Surat"]
```

step 5 Apply Multiple Conditions

```python
# Filter IT, Salary > 50k.
df[(df["Dept"]=="IT") & (df["Salary"]>50000)]
```

step 6 Sort by Salary

```python
# Sort by Salary ascending.
df.sort_values(by="Salary")
```

step 7 Sort Salary from Highest to Lowest

```python
# Sort by Salary descending.
df.sort_values(by="Salary", ascending=False)
```

step 8 Sort by Salary and Department

```python
# Sort by Dept, then Salary.
df.sort_values(by=["Dept","Salary"], ascending=[True,False])
```

step 9 Aggregation Operations

```python
# Sum of all salaries.
df["Salary"].sum()
```

step 10 Calculate Average Salary

```python
# Average salary.
df["Salary"].mean()
```

step 11 Find Highest Salary

```python
# Max salary.
df["Salary"].max()
```

step 12 Find Lowest Salary

```python
# Min salary.
df["Salary"].min()
```

step 13 Count Employees

```python
# Count employees.
df["ID"].count()
```

step 14 Average Salary by Department

```python
# Avg Salary by Dept.
df.groupby("Dept")["Salary"].mean()
```

step 15 Minimum Salary by Department

```python
# Min Salary by Dept.
df.groupby("Dept")["Salary"].min()
```

step 16 Employee Count by Department

```python
# Employee count by Dept.
df.groupby("Dept")["ID"].count()
```

step 17 Total Salary Dept Wise

```python
# Total Salary by Dept.
df.groupby("Dept")["Salary"].sum()
```

step 18 Maximum Salary by Department

```python
# Max Salary by Dept.
df.groupby("Dept")["Salary"].max()
```

step 19 Average Salary by City

```python
# Avg Salary by City.
df.groupby("City")["Salary"].mean()
```

step 20  Multiple Aggregation Functions

```python
# Multi-agg by Dept.
df.groupby("Dept").agg({"Salary":["count","mean","max","min","sum"]})
```

step 21 Save DataFrame to CSV

```python
# Save to CSV.
df.to_csv("employee_data.csv", index=False)
from google.colab import files
files.download("employee_data.csv")
```
